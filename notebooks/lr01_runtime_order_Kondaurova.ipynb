В этом notebook, я создаю текстовую переменную, вывожу ее и проверяю, что происходит с runtime после перезапуска.

```python
project_title = "Text Lab"
```

```python
print(project_title)
```

##Что произошло?
В ячейке 2 я создала переменную "project_title". Это изменило состояние runtime и теперь в памяти есть эта переменная. В ячейке 3 я ввела ее через "print()", что означает хранение переменных в runtime

```python
print(эники бэники)
```

##Рабочий и ошибочный случай
В рабочем случае в ячейках 2 и 3 переменная создана и выведена, поэтому работает.
В ошибочном случае я ввела "эники бэники", этой переменной нет в runtime, так что в результате мы получаем ошибку.
##Вывод
Runtime хранит только те переменные, которые мы создали и запустили. Если переменной нет в runtime код не работает.

```python
#Собственный тест: переименовываю переменную и проверяю, обновился ли runtime
project_title = "Новое имя"
print(project_title)
```

##Собственный тест
Я перезаписала "project_title" новым значением и вывела его. Вывод изменился, что показывает - runtime обновил значение переменной.

##Проверка после restart
Сейчас перезапущу runtime и нажму Run all, чтобы проверить пройдет ли notebook кроме ячейки 5 без ошибок.

##Итог
Notebook это файл с ячейками, а runtime это временная память.
После restart все переменные исчезают.
Порядок расположения важнее расположения на страницe.
Restart и Run all выступают в качестве теста на воспроизводимость кодов.
